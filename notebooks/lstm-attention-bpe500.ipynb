# LSTM with attention on BPE

The LSTM-with-attention run on 500 BPE tokens instead of characters, the best n-gram tokenization. BPC is still per character, so it compares directly with the character runs. It trains in fp32 with batches of 64 (learning rate 2e-3), like the character run, since under fp16 autocast it stalled at the unigram level.

Attach to **Serverless GPU, A10**. `tracked_run` logs every step and epoch to the `lstm` MLflow experiment, saves the best-validation checkpoint to the volume and logs sampled turns.
