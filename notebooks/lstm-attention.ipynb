# LSTM with attention

Causal self-attention with 4 heads over the top of three LSTM layers; hidden size 480 (5.75M parameters) is the closest to the baseline's 5.5M whose head size suits the fused CUDA attention kernels. It trains in fp32, since under fp16 autocast it stalled at the unigram level, and fp32 needs batches of 64 (learning rate 2e-3) to fit on the A10.

Attach to **Serverless GPU, A10**. `tracked_run` logs every step and epoch to the `lstm` MLflow experiment, saves the best-validation checkpoint to the volume and logs sampled turns.
