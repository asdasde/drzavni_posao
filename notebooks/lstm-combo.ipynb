# LSTM with residuals, locked dropout and attention

Pre-norm residual connections around every LSTM layer, one dropout mask per sequence, and 4-head causal self-attention on top; hidden size 448 keeps it near 5.5M parameters. It trains in fp32 with batches of 64 (learning rate 2e-3), since fp16 autocast stalled the other attention LSTMs at the unigram level.

Attach to **Serverless GPU, A10**. `tracked_run` logs every step and epoch to the `lstm` MLflow experiment, saves the best-validation checkpoint to the volume and logs sampled turns.
